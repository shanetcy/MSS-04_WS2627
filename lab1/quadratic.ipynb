```python
def q_solve(a,b,c):
    D = (b**2)-(4*a*c)
    if D >= 0:
        x1 = ((-b)+(D**0.5))/(2*a)
        x2 = ((-b)-(D**0.5))/(2*a)
        if x1==x2:
            return x1
    else:
        re = (-b)/(2*a)
        im = (((-D)**0.5)/(2*a))*1j
        x1 = re+im
        x2 = re-im
    return x1,x2
```

```python
q_solve(1,2,5)
```

```python
import cmath
```

```python
def q_solve_complex(a,b,c):
    D = (b**2)-(4*a*c)
    x1 = ((-b)+(cmath.sqrt(D)))/(2*a)
    x2 = ((-b)-(cmath.sqrt(D)))/(2*a)
    if x1==x2:
        return x1
    return x1,x2
```

```python
q_solve_complex(1,2,5)
```

```python
q_solve_complex(1,-2j,-1)
```

```python
q_solve_complex(1,-3j,-2)
```
