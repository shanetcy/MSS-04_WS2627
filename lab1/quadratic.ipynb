```python
def q_solve(a,b,c):
    D = (b**2)-(4*a*c)
    if D >= 0:
        x1 = ((-b)+(D**0.5))/(2*a)
        x2 = ((-b)-(D**0.5))/(2*a)
        if x1==x2:
            return x1
    else:
        re = (-b)/(2*a)
        im = (-D) ** 0.5
        x1 = complex(re,im)
        x2 = complex(re,-im)
    return x1,x2
```
