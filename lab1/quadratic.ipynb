```python
def q_solve(a,b,c):
    D = (b**2)-(4*a*c)
    if D >= 0:
        x1 = ((-b) + (D ** 0.5))/(2*a)
        x2 = ((-b) - (D ** 0.5))/(2*a)
        if x1 == x2:
            return x1
        return x1,x2
    return "no real solution"
```

```python
q_solve(1,-2,1)
```

```python
q_solve(1,-3,2)
```

```python
q_solve(1,2,5)
```
